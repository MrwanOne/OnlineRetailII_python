```python
from pathlib import Path
import sys
sys.path.insert(0, str(Path.cwd().parent / "src"))
from load_data import load_online_retail
from clean_data import basic_clean
from analysis import monthly_sales, top_products

df_raw = load_online_retail()
print("Shape:", df_raw.shape)
display(df_raw.head())
```

## 1. Data audit

```python
df_raw.info()
display(df_raw.isna().sum().sort_values(ascending=False).to_frame("missing_count"))
print("Exact duplicate rows:", df_raw.duplicated().sum())
```

## 2. Basic cleaning

```python
df = basic_clean(df_raw)
display(df.head())
```

## 3. Monthly sales trend

```python
import matplotlib.pyplot as plt
monthly = monthly_sales(df)
ax = monthly.plot(x="InvoiceDate", y="Revenue", figsize=(12, 5), legend=False)
ax.set_title("Monthly line revenue — document cancellation/return policy")
ax.set_xlabel("Month")
ax.set_ylabel("Revenue")
plt.tight_layout()
plt.show()
```

## 4. Top products by recorded revenue

```python
products = top_products(df, n=10)
display(products)
products.sort_values("Revenue").plot.barh(x="Description", y="Revenue", figsize=(10, 6), legend=False)
plt.title("Top 10 products by recorded line revenue")
plt.tight_layout()
plt.show()
```
